```python
# Question 1

import math

# Givens
n_old = 24
mean = 32.46
old_std_dev = 6.92
new_values = [23.5, 30.5, 42.3, 34.5, 49.5, 44.2]

# New Mean
n_new = n_old + len(new_values)
new_mean = (mean * n_old +sum(new_values))/n_new

# New Standard Deviation

sum_squares_old = (old_std_dev ** 2) * (n_old - 1)
sum_squares_new = sum((x - new_mean) ** 2 for x in new_values)
std_new = math.sqrt((sum_squares_old + sum_squares_new) / (n_new - 1))

print(f"Updated Mean: {new_mean:.2f}")
print(f"Updated Standard Deviation: {std_new:.2f}")
```
