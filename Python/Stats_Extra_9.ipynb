```python
import numpy as np
import scipy.stats as stats
from scipy.special import comb

# Question 1
p_values = np.linspace(0.7, 0.9, 20)
n, k, total_seats = 43, 40, 40  # Total passengers, occupied seats, and capacity

def prob_exactly_three_empty(p):
    return stats.binom.pmf(37, n, p)  # Probability of exactly 37 showing up

def prob_at_least_one_empty(p):
    return sum(stats.binom.pmf(i, n, p) for i in range(0, 41))

prob_three_empty = [prob_exactly_three_empty(p) for p in p_values]
p_max_prob = p_values[np.argmax(prob_three_empty)]
print(p_max_prob)

prob_one_or_more_empty = [prob_at_least_one_empty(p) for p in p_values]
p_less_95 = [p for p, prob in zip(p_values, prob_one_or_more_empty) if prob < 0.95]
print(p_less_95)
```

```python
# Question 2: Hypergeometric Distribution
N, D, n = 10, 3, 4  # Total missiles, defective missiles, drawn samples
prob_all_fire = comb(7, 4) / comb(10, 4)  # P(X=4)
prob_at_most_2_not_fire = sum(comb(7, i) * comb(3, 4 - i) / comb(10, 4) for i in range(2, 5))

print(prob_all_fire)
print(prob_at_most_2_not_fire)
```

```python
# Question 3: Poisson Distribution
lambda_per_acre = 12
prob_fewer_than_7 = sum(stats.poisson.pmf(i, lambda_per_acre) for i in range(7))
print(prob_fewer_than_7)
```

```python
# Question 4: Binomial Distribution
p_protected = 0.65
prob_none_disease = stats.binom.pmf(0, 5, 1 - p_protected)
prob_fewer_2 = sum(stats.binom.pmf(i, 5, 1 - p_protected) for i in range(2))
prob_more_3 = sum(stats.binom.pmf(i, 5, 1 - p_protected) for i in range(4, 6))
print(prob_none_disease)
print(prob_fewer_2)
print(prob_more_3)
```

```python
# Question 5: Poisson Process
lambda_aircraft = 6
prob_exactly_4 = stats.poisson.pmf(4, lambda_aircraft)
prob_at_least_4 = sum(stats.poisson.pmf(i, lambda_aircraft) for i in range(4, 100))
lambda_12hr = 6 * 12
prob_75_arrivals = stats.poisson.pmf(75, lambda_12hr)
print(prob_exactly_4)
print(prob_at_least_4)
print(prob_75_arrivals)
```

```python
# Question 6: Binomial Distribution for Truck Failures
p_fail = 0.25
prob_3_to_6 = sum(stats.binom.pmf(i, 15, p_fail) for i in range(3, 7))
prob_fewer_4 = sum(stats.binom.pmf(i, 15, p_fail) for i in range(4))
prob_more_5 = sum(stats.binom.pmf(i, 15, p_fail) for i in range(6, 16))
print(prob_3_to_6)
print(prob_fewer_4)
print(prob_more_5)
```

```python
# Question 7: Geometric and Negative Binomial Distributions
p_strike = 0.2
prob_first_strike_3 = (1 - p_strike) ** 2 * p_strike  # Geometric
prob_third_strike_7 = stats.nbinom.pmf(7 - 3, 3, p_strike)  # Negative Binomial
mean_drills = 3 / p_strike
var_drills = (3 * (1 - p_strike)) / (p_strike ** 2)
print(prob_first_strike_3)
print(prob_third_strike_7)
print(mean_drills)
print(var_drills)
```
