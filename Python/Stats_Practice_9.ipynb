```python
# Import required libraries
import numpy as np
from scipy import stats

alpha = 0.05
```

```python
# 1.


data1 = np.array([2.5, 3.0, 3.1, 4.0, 1.2, 5.0, 4.1, 3.9, 3.2, 3.3, 2.8, 4.1, 2.7, 2.9, 3.7])
mu0 = 3  # current yield

# Hypotheses
print("H0: mu = 3")
print("H1: mu > 3")

# One-sided t-test
t_stat, p_value_two_sided = stats.ttest_1samp(data1, mu0)

# Convert to one-sided p-value
p_value = p_value_two_sided / 2 if t_stat > 0 else 1 - (p_value_two_sided / 2)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < alpha:
    print("Reject H0: Fertilizer increases yield\n")
else:
    print("Fail to reject H0: No significant increase\n")
```

```python
# 2.


S1 = np.array([69, 75, 79, 80, 81, 82, 86, 89, 91, 94, 97])
S2 = np.array([59, 62, 66, 70, 70, 75, 76, 77, 78, 79, 81, 84, 85, 86, 94])

print("H0: mu1 = mu2")
print("H1: mu1 != mu2")

t_stat, p_value = stats.ttest_ind(S1, S2, equal_var=True)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < alpha:
    print("Reject H0: Means are significantly different\n")
else:
    print("Fail to reject H0: No significant difference\n")
```

```python
# 3.


W1 = np.array([84, 73, 79, 80, 81, 82, 86, 89, 91, 94,
               86, 72, 62, 54, 77, 63, 85, 59, 66, 79])

W2 = np.array([59, 62, 66, 70, 75, 76, 77, 78, 79, 81])

print("H0: mu1 <= mu2")
print("H1: mu1 > mu2")

t_stat, p_value_two_sided = stats.ttest_ind(W1, W2, equal_var=False)

# Convert to one-sided
p_value = p_value_two_sided / 2 if t_stat > 0 else 1 - (p_value_two_sided / 2)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < alpha:
    print("Reject H0: W1 has higher sodium\n")
else:
    print("Fail to reject H0: Not enough evidence\n")

```

```python
# 4.


pre = np.array([10.80, 12.90, 9.59, 8.81, 12.00, 6.07])
post = np.array([10.60, 16.60, 17.20, 14.00, 10.60, 8.60])

print("H0: mu_diff = 0")
print("H1: mu_diff != 0")

t_stat, p_value = stats.ttest_rel(pre, post)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < alpha:
    print("Reject H0: Significant change after surgery\n")
else:
    print("Fail to reject H0: No significant change\n")

```

```python
# 5


# Urban
x1, n1 = 20, 200

# Rural
x2, n2 = 10, 150

p1 = x1 / n1
p2 = x2 / n2

print("H0: p1 <= p2")
print("H1: p1 > p2")

# pooled proportion
p_pool = (x1 + x2) / (n1 + n2)

# standard error
se = np.sqrt(p_pool * (1 - p_pool) * (1/n1 + 1/n2))

# z statistic
z_stat = (p1 - p2) / se

# one-sided p-value
p_value = 1 - stats.norm.cdf(z_stat)

print("z-statistic:", z_stat)
print("p-value:", p_value)

if p_value < alpha:
    print("Reject H0: Higher breast cancer rate in urban area\n")
else:
    print("Fail to reject H0: Not enough evidence\n")
```
