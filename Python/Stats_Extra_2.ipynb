```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
math = r"C:\Users\dunca\OneDrive\Desktop\STAT 603\HW2\Problem1.14.student-mat.csv"
por = r"C:\Users\dunca\OneDrive\Desktop\STAT 603\HW2\Problem1.14.student-por.csv"

math_students = pd.read_csv(math, sep=';')
por_students = pd.read_csv(por, sep=';')

# Add a column to distinguish datasets
math_students["course"] = "Math"
por_students["course"] = "Portuguese"

# Combine datasets
data = pd.concat([math_students, por_students])

# (a) Conditional Histograms: Weekday Alcohol Consumption
plt.figure(figsize=(10, 5))
sns.histplot(data=data, x="Dalc", hue="course", multiple="dodge", bins=5, shrink=0.8)
plt.xlabel("Weekday Alcohol Consumption (Dalc)")
plt.ylabel("Count")
plt.title("Weekday Alcohol Consumption by Course")
plt.show()

# (b) Conditional Histograms: Weekend Alcohol Consumption by Family Size
plt.figure(figsize=(10, 5))
sns.histplot(data=data, x="Walc", hue="famsize", multiple="dodge", bins=5, shrink=0.8)
plt.xlabel("Weekend Alcohol Consumption (Walc)")
plt.ylabel("Count")
plt.title("Weekend Alcohol Consumption by Family Size")
plt.show()

# (c) Box Plots: Total Alcohol Consumption by Student Type
data["Total_Alcohol"] = data["Dalc"] + data["Walc"]
data["Student_Type"] = data["school"] + "_" + data["sex"] + "_" + data["famsize"] + "_" + data["romantic"]

plt.figure(figsize=(12, 6))
sns.boxplot(data=data, x="Student_Type", y="Total_Alcohol")
plt.xticks(rotation=90)
plt.xlabel("Student Type")
plt.ylabel("Total Alcohol Consumption")
plt.title("Total Alcohol Consumption by Student Type")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
file = r"C:\Users\dunca\OneDrive\Desktop\STAT 603\HW2\Problem1.15.credit-card-default-data.xlsx"
data = pd.read_excel(file, header=1, index_col=0)

# (a) Conditional Histograms: Debt (X1) by Default Status
plt.figure(figsize=(10, 5))
sns.histplot(data=data, x="LIMIT_BAL", hue="default payment next month", multiple="dodge", bins=30, log_scale=True, shrink=0.8)
plt.xlabel("Debt (LIMIT_BAL)")
plt.ylabel("Count")
plt.title("Debt Distribution by Default Status")
plt.show()

# (b) Box Plots: Effect of Gender, Education, and Marital Status on Debt
plt.figure(figsize=(12, 5))
sns.boxplot(data=data, x="SEX", y="LIMIT_BAL")
plt.xlabel("Gender (1=Male, 2=Female)")
plt.ylabel("Debt (LIMIT_BAL)")
plt.title("Debt by Gender")
plt.show()

plt.figure(figsize=(12, 5))
sns.boxplot(data=data, x="EDUCATION", y="LIMIT_BAL")
plt.xlabel("Education Level")
plt.ylabel("Debt (LIMIT_BAL)")
plt.title("Debt by Education Level")
plt.show()

plt.figure(figsize=(12, 5))
sns.boxplot(data=data, x="MARRIAGE", y="LIMIT_BAL")
plt.xlabel("Marital Status (1=Married, 2=Single, 3=Others)")
plt.ylabel("Debt (LIMIT_BAL)")
plt.title("Debt by Marital Status")
plt.show()
```
