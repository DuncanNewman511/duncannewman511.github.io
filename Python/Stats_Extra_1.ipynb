```python
import pandas as pd
import numpy as np
import seaborn as sns
from scipy.stats import norm
import matplotlib.pyplot as plt

data = r'C:\Users\dunca\OneDrive\Desktop\STAT 603\Cars93.csv'
cars_df = pd.read_csv(data)
print(cars_df.head())
```

```python
# Plot for the catagorical variable "Manufacturer"

plt.figure(figsize = (12,6))
cars_df['Manufacturer'].value_counts().plot(kind='bar', color='red')
plt.title('Number of Cars per Manufacturer')
plt.xlabel('Manufacturer')
plt.ylabel('Count')
plt.xticks(rotation=90)
plt.show
```

```python
# Plot for the catagorical variable "Type"

plt.figure(figsize = (12,6))
cars_df['Type'].value_counts().plot(kind='bar', color='green')
plt.title('Number of Cars per Style of Car')
plt.xlabel('Body Style')
plt.ylabel('Count')
plt.xticks(rotation=90)
plt.show
```

```python
# The conclusions I can make from these graphs is that the most 
# popular car based on this data is a midsize car from either Chevrolet or Ford. 
```

```python
# Create density histograms for the variables Min.Price, Max.Price, Weight, and Length 
# using a different color for each histogram using an appropriate number of bins.

# Create variables and colors
variables = ['Min.Price', 'Max.Price', 'Weight', 'Length']
colors = ['red', 'blue', 'green', 'purple']

# Function to plot density histograms w/normal density curves 
# that loops through variables and colors dictionary from the dataframe
for var, color in zip(variables,colors):
    plt.figure(figsize=(8,4))
    sns.histplot(cars_df[var], bins=10,kde=False, color=color, stat='density', edgecolor='black')
    
    # Fit the normal curve to the data
    mu, std = norm.fit(cars_df[var].dropna())

    # Plot the normal density curve
    xmin, xmax = plt.xlim()
    x = np.linspace(xmin, xmax, 100)
    p = norm.pdf(x, mu, std)
    plt.plot(x, p, 'k', linewidth=2)
    
    plt.title(f'Density Histogram of {var} with Normal Density Curve')
    plt.xlabel(var)
    plt.ylabel('Density')
    plt.show()
```
