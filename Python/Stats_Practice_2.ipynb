```python
# Problem 1b

import numpy as np
from scipy.optimize import minimize

# data
x = np.array([7, 10, 7, 8, 9, 10, 9, 8, 11, 12, 14, 15, 12, 6, 23])

def neg_log_likelihood_theta(theta):
    if theta <= 0:
        return np.inf
    return -np.sum(np.log((1/(2*theta**3)) * x**2 * np.exp(-x/theta)))

# Method 1: BFGS
res_bfgs = minimize(lambda t: neg_log_likelihood_theta(t[0]), x0=[1.0], method='BFGS')
theta_hat_bfgs = res_bfgs.x[0]

# Method 2: Nelder-Mead
res_nm = minimize(lambda t: neg_log_likelihood_theta(t[0]), x0=[1.0], method='Nelder-Mead')
theta_hat_nm = res_nm.x[0]

print(theta_hat_bfgs, theta_hat_nm)
```

```python
# Problem 2b
from scipy.optimize import root

y = np.array([6, 5, 8, 8, 13, 11, 7, 8, 7, 10, 8, 4, 12, 3, 11, 9, 15, 12, 6, 3])
m = len(y)

def score_lambda(lam):
    if lam <= 0:
        return np.inf
    return np.sum(y)/lam - m - m*(np.exp(-lam)/(1-np.exp(-lam)))

root_res = root(lambda lam: score_lambda(lam[0]), x0=[5.0])
lambda_hat = root_res.x[0]
print(lambda_hat)
```

```python
# Problem 3c

from scipy.optimize import minimize

# Data
data = np.array([1, 2, 1, 1, 1, 1, 3, 1, 5, 1])
n = len(data)
n1 = np.sum(data == 1)
cK = np.max(data)

# Negative log-likelihood
def neg_log_likelihood_theta(theta):
    theta = theta[0]
    return -(n1*np.log(theta) + (n - n1)*np.log((1-theta)/(cK-1)))

# Use bounded optimization to keep 0 < theta < 1
res = minimize(
    neg_log_likelihood_theta,
    x0=[0.5],
    bounds=[(1e-6, 1-1e-6)],
    method='L-BFGS-B'
)

theta_hat = res.x[0]
print(theta_hat)
```
