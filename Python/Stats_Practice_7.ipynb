```python
import numpy as np
import scipy.stats as stats
import pandas as pd
```

```python
# 1
data1 = np.array([1.9, 2.4, 3.0, 3.5, 4.2])
n1 = len(data1)
s1 = np.std(data1, ddof=1)
alpha = 0.10

chi2_lower = stats.chi2.ppf(alpha/2, df=n1-1)
chi2_upper = stats.chi2.ppf(1 - alpha/2, df=n1-1)

sigma_lower = np.sqrt((n1-1)*s1**2 / chi2_upper)
sigma_upper = np.sqrt((n1-1)*s1**2 / chi2_lower)

print("Q1 CI for sigma:", (sigma_lower, sigma_upper))
```

```python
# 2
A = np.array([3.5,2.7,3.9,4.2,3.6,2.7,3.3,5.2,4.2,2.9,4.4,5.2,4.0,4.1,3.4])
B = np.array([4.7,3.9,4.5,5.5,4.0,5.3,4.3,6.0,5.2,3.7,5.5,6.2,5.1,5.4,4.8])

nA, nB = len(A), len(B)
meanA, meanB = np.mean(A), np.mean(B)
varA, varB = np.var(A, ddof=1), np.var(B, ddof=1)

sp2 = ((nA-1)*varA + (nB-1)*varB) / (nA+nB-2)
sp = np.sqrt(sp2)

alpha = 0.01
tcrit = stats.t.ppf(1-alpha/2, df=nA+nB-2)

margin = tcrit * sp * np.sqrt(1/nA + 1/nB)
ci2 = ((meanA - meanB) - margin, (meanA - meanB) + margin)

print("Q2 CI:", ci2)
```

```python
# 3
n1, x1, s1 = 8, 1.98, 0.51
n2, x2, s2 = 13, 1.30, 0.35

se = np.sqrt(s1**2/n1 + s2**2/n2)

df = (s1**2/n1 + s2**2/n2)**2 / ((s1**2/n1)**2/(n1-1) + (s2**2/n2)**2/(n2-1))

alpha = 0.05
tcrit = stats.t.ppf(1-alpha/2, df)

margin = tcrit * se
ci3 = ((x1 - x2) - margin, (x1 - x2) + margin)

print("Q3 CI:", ci3)
```

```python
# 4a
T1 = np.array([5.613,6.130,5.422,5.947,5.514,5.322,5.690,5.243,5.920,5.859])
T2 = np.array([5.934,5.335,5.826,4.821,5.664,5.292,5.257,6.245,5.981,5.197])

n1, n2 = len(T1), len(T2)
s1_sq, s2_sq = np.var(T1, ddof=1), np.var(T2, ddof=1)

alpha = 0.10
F_lower = stats.f.ppf(alpha/2, n1-1, n2-1)
F_upper = stats.f.ppf(1-alpha/2, n1-1, n2-1)

ratio_lower = (s1_sq/s2_sq) / F_upper
ratio_upper = (s1_sq/s2_sq) / F_lower

print("Q4a CI variance ratio:", (ratio_lower, ratio_upper))


# 4b
sp2 = ((n1-1)*s1_sq + (n2-1)*s2_sq) / (n1+n2-2)
sp = np.sqrt(sp2)

tcrit = stats.t.ppf(1-alpha/2, n1+n2-2)
margin = tcrit * sp * np.sqrt(1/n1 + 1/n2)

ci4b = ((np.mean(T1) - np.mean(T2)) - margin,
        (np.mean(T1) - np.mean(T2)) + margin)

print("Q4b CI:", ci4b)
```

```python
# 5
df = pd.read_csv("glucoselevels.csv")
d = df.iloc[:,0] - df.iloc[:,1]

n = len(d)
mean_d = np.mean(d)
sd_d = np.std(d, ddof=1)

tcrit = stats.t.ppf(1-0.05/2, n-1)
margin = tcrit * sd_d / np.sqrt(n)

ci5 = (mean_d - margin, mean_d + margin)
print("Q5 CI:", ci5)
```

```python
# 6
n1, x1 = 500, 98
n2, x2 = 500, 120

p1, p2 = x1/n1, x2/n2

alpha = 0.10
z = stats.norm.ppf(1-alpha/2)

se = np.sqrt(p1*(1-p1)/n1 + p2*(1-p2)/n2)

ci6 = ((p1 - p2) - z*se, (p1 - p2) + z*se)
print("Q6 CI:", ci6)
```

```python
# 7
n1, x1 = 100, 10
n2, x2 = 100, 6

p1, p2 = x1/n1, x2/n2

alpha = 0.02
z = stats.norm.ppf(1-alpha/2)

se = np.sqrt(p1*(1-p1)/n1 + p2*(1-p2)/n2)

ci7 = ((p1 - p2) - z*se, (p1 - p2) + z*se)
print("Q7 CI:", ci7)
```

```python
# 8
data8 = np.array([6,5,8,8,13,11,7,8,7,10,8,4,12,3,11,9,15,12,6,3])

alpha_prior = 2
beta_prior = 3

alpha_post = alpha_prior + np.sum(data8)
beta_post = beta_prior + len(data8)

# 95% credible interval
lower = stats.gamma.ppf(0.025, a=alpha_post, scale=1/beta_post)
upper = stats.gamma.ppf(0.975, a=alpha_post, scale=1/beta_post)

print("Q8 Posterior: Gamma({}, {})".format(alpha_post, beta_post))
print("Q8 CI:", (lower, upper))
```
