```python
# Import necessary libraries
import numpy as np
from scipy.stats import norm, gamma
from math import comb

# Problem 1: Credit card debt
mean_debt = 3262
std_debt = 1100

# (a) P(X >= 1000)
p1a = 1 - norm.cdf(1000, loc=mean_debt, scale=std_debt)

# (b) P(X > 4000)
p1b = 1 - norm.cdf(4000, loc=mean_debt, scale=std_debt)

# (c) P(3000 <= X <= 4000)
p1c = norm.cdf(4000, loc=mean_debt, scale=std_debt) - norm.cdf(3000, loc=mean_debt, scale=std_debt)

# Problem 2: Charitable contributions
mean_contrib = 792
std_contrib = 103

p2_lower = norm.ppf(0.25, loc=mean_contrib, scale=std_contrib)
p2_upper = norm.ppf(0.75, loc=mean_contrib, scale=std_contrib)

# Problem 3: Competency test
mean_score = 400
std_score = 100

# (a) Top 3%
p3a = norm.ppf(0.97, loc=mean_score, scale=std_score)

# (b) Bottom 1.5%
p3b = norm.ppf(0.015, loc=mean_score, scale=std_score)

# Problem 4: Gamma distribution
mean_gamma = 10
var_gamma = 50
beta = var_gamma / mean_gamma
alpha = mean_gamma / beta

# (b) P(X <= 50)
p4b = gamma.cdf(50, a=alpha, scale=beta)

# (c) P(X <= 10)
p4c = gamma.cdf(10, a=alpha, scale=beta)

# Problem 5: Committee selection
total_members = 12
total_committee = 4

# (a) All nationalities represented (1 each of Canadian, Japanese, Italian, German)
p5a = comb(2,1) * comb(3,1) * comb(5,1) * comb(2,1) / comb(total_members, total_committee)

# (b) All except Italian (1 Canadian, 1 Japanese, 1 German + 1 from any of those 3)
ways_to_choose_3_types = comb(2,1) * comb(3,1) * comb(2,1)
remaining = comb(7 - 3, 1)  # pick 1 more from the remaining non-Italians
p5b = ways_to_choose_3_types * remaining / comb(total_members, total_committee)

# Problem 6: Light bulb lifetime (binomial-like combinatorics)
ways = comb(10,4) * comb(6,2)
p_comb = (0.5**4) * (0.2**2) * (0.3**4)
p6 = ways * p_comb

# Display results
results = {
    "Problem 1a (>= $1000)": p1a,
    "Problem 1b (> $4000)": p1b,
    "Problem 1c ($3000 - $4000)": p1c,
    "Problem 2 (Middle 50%)": (p2_lower, p2_upper),
    "Problem 3a (Top 3%)": p3a,
    "Problem 3b (Above Bottom 1.5%)": p3b,
    "Problem 4a (Alpha, Beta)": (alpha, beta),
    "Problem 4b (≤ 50 weeks)": p4b,
    "Problem 4c (≤ 10 weeks)": p4c,
    "Problem 5a (All represented)": p5a,
    "Problem 5b (All except Italian)": p5b,
    "Problem 6 (Exact combination)": p6
}

results
```
