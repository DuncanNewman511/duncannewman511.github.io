# 6. Secondary Observations

The next obervation for a player on a roster sheet would be two things. The first being which hand do you throw with, and the other would be which side of the plate do you throw with. The first is a very important metric because of where you could possibly play in the field. For example, it is counterintuitive to use a left-handed thrower as catcher, because he is at a disadvantage for most of his throws given the angles he will have to throw from. But, left-handed throwers have an advantage at first base, given the angles that are played over in that position are better for them. and interesting observation to see is the appeearance of the "Both" catagory. There is one player to do it, a pitcher named Pat Venditte, an ambidextrous pitcher that founded the "Venditte Rule". The second is important because depending on whether the opposing pitcher is left or right-handed, you need a ceratin amount of both sides in order to balance out your batting lineup.

In order to do this, I have created a histograph of each handedness throwing and hitting. I made sure to format it as such that they are vertically related to each other, so for easier reading, as well as highliting in a salmon color  which catagory I would fall into, to again compare myself to the dataset itself.

```python
# Showing amount of right handed hitter and left handed hitters and switch
plt.figure(figsize=(12, 8))
ax = sb.histplot(MLB_db["bats"], kde = True)
ax.patches[0].set_facecolor('salmon')
plt.title("Batting Stance All Players")
plt.xlabel("Bats")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(12,8))
ax = sb.histplot(HOF_db["bats"], kde = True)
ax.patches[0].set_facecolor('salmon')
plt.title("Batting Stance of HOF's")
plt.xlabel("Bats")
plt.ylabel("Frequency")
plt.show()

# Showing amount of right handed thrower and left handed thrower
plt.figure(figsize=(12,8))
ax = sb.histplot(MLB_db["throws"], kde = True)
ax.patches[0].set_facecolor('salmon')
plt.title("Throwing Hand of All Players")
plt.xlabel("Throws")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(12,8))
ax = sb.histplot(HOF_db["throws"], kde = True)
ax.patches[0].set_facecolor('salmon')
plt.title("Throwing Hand of HOF'rs")
plt.xlabel("Throws")
plt.ylabel("Frequency")
plt.show()
```

# 8. Final Observation

The final observation I had made was really what this has been all about, and that is answering the most important question. How much does each factor matter in order to take yourself from a professional player, to being in the Hall of Fame? Is there a formula that correlates whether or not a professional player will be in the Hall of Fame, or is there truly not a "one size fits all" approach. Below you will see this observation. As you can see, there isn't a strong correlation to the things that would matter on the surface of the sport. This being measurable, like height and weight. Where both of these values are around the 0.5 area, meaning they aren't fully correlated to each other. This would then mean that this proves the theory that there really is a true cookie cutter frame for a baseball player to be in inorder to make the Hall of Fame, but more that the skill of the player matters more.

In order to do this, I have included two heatmaps, the first being with all of the MLB players, and then second being with just the HOF players. I have included a gradient color scheme to truly represent a "heatmap" so to speak.

```python
# Correlation Heatmaps

numeric_db = MLB_db.select_dtypes(include=[np.number])
plt.figure(figsize=(12, 8))
sb.heatmap(numeric_db.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation of MLB Heatmap')
plt.show()

numeric_db = HOF_db.select_dtypes(include=[np.number])
plt.figure(figsize=(12, 8))
sb.heatmap(numeric_db.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation of HOF Heatmap')
plt.show()
```
