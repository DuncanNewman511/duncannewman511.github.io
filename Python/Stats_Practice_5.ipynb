```python
# STAT 604 HW5 Duncan Newman

import numpy as np
import pandas as pd
from scipy import stats
```

```python
# 6.4 The Weight of Rats

# Given data
n = 40
mean = 340
std = 75

# Standard error
se = std / np.sqrt(n)

# (a) 68% confidence interval (z ≈ 1)
z_68 = stats.norm.ppf(0.84)  # 68% CI -> 0.16 in each tail
ci_68 = (mean - z_68 * se, mean + z_68 * se)

print("68% CI:", ci_68)

# (b) 99% confidence interval
z_99 = stats.norm.ppf(0.995)
ci_99 = (mean - z_99 * se, mean + z_99 * se)

print("99% CI:", ci_99)
```

```python
# 6.5 The Weight of Mice

# Data
weights = np.array([21, 23, 27, 19, 17, 18, 20, 15, 17, 22])

n = len(weights)
mean = np.mean(weights)
std = np.std(weights, ddof=1)
se = std / np.sqrt(n)

# (a) 80% confidence interval
t_80 = stats.t.ppf(0.90, df=n-1)
ci_80 = (mean - t_80 * se, mean + t_80 * se)

print("80% CI:", ci_80)

# (b) 95% confidence interval
t_95 = stats.t.ppf(0.975, df=n-1)
ci_95 = (mean - t_95 * se, mean + t_95 * se)

print("95% CI:", ci_95)
```

```python
# 6.6 Probability of a Female Birth

n = 2009
girls = 1026
boys = 983

p_hat = girls / n
se = np.sqrt(p_hat * (1 - p_hat) / n)

# (a) 99% CI for female probability
z_99 = stats.norm.ppf(0.995)
ci_female = (p_hat - z_99 * se, p_hat + z_99 * se)

print("99% CI for female probability:", ci_female)

# (b) 99% CI for male probability
p_male = boys / n
se_male = np.sqrt(p_male * (1 - p_male) / n)
ci_male = (p_male - z_99 * se_male, p_male + z_99 * se_male)

print("99% CI for male probability:", ci_male)

# (c) Check overlap
overlap = not (ci_female[1] < ci_male[0] or ci_male[1] < ci_female[0])
print("Do intervals overlap?", overlap)
```

```python
# 6.7 Breast Tissue Dataset

df = pd.read_excel("BreastTissue.xls")

# Extract correct columns
class_col = df.columns[1]   # Class
i0_col = df.columns[2]      # I0

# Clean class labels
df[class_col] = df[class_col].astype(str).str.strip().str.lower()

# Convert I0 column to numeric (fixes string issue)
df[i0_col] = pd.to_numeric(df[i0_col], errors="coerce")

# Drop any rows where I0 could not be converted
df = df.dropna(subset=[i0_col])

# Filter carcinoma and adipose
carcinoma = df[df[class_col] == "carcinoma"][i0_col]
adipose = df[df[class_col] == "adipose"][i0_col]

def confidence_interval(data, confidence=0.99):
    n = len(data)
    mean = np.mean(data)
    std = np.std(data, ddof=1)
    se = std / np.sqrt(n)
    t_value = stats.t.ppf((1 + confidence) / 2., df=n - 1)
    return (mean - t_value * se, mean + t_value * se)

# (a) 99% CI for Carcinoma I0
ci_carcinoma = confidence_interval(carcinoma)
print("99% CI for Carcinoma (I0):", ci_carcinoma)

# (b) 99% CI for Adipose I0
ci_adipose = confidence_interval(adipose)
print("99% CI for Adipose (I0):", ci_adipose)

# (c) Check overlap
overlap = not (ci_carcinoma[1] < ci_adipose[0] or ci_adipose[1] < ci_carcinoma[0])
print("Do the intervals overlap?", overlap)
```

```python
# 6.8 Wine Dataset

df = pd.read_csv("wine.data", header=None)

# Column names from dataset description
df.columns = ['Class', 'Alcohol', 'Malic_acid', 'Ash', 'Alcalinity',
              'Magnesium', 'Total_phenols', 'Flavanoids',
              'Nonflavanoid_phenols', 'Proanthocyanins',
              'Color_intensity', 'Hue',
              'OD280_OD315', 'Proline']

# Filter region 1 and region 3
region1 = df[df['Class'] == 1]['Flavanoids']
region3 = df[df['Class'] == 3]['Flavanoids']

def confidence_interval(data, confidence=0.99):
    n = len(data)
    mean = np.mean(data)
    std = np.std(data, ddof=1)
    se = std / np.sqrt(n)
    t_val = stats.t.ppf((1 + confidence) / 2., df=n-1)
    return (mean - t_val * se, mean + t_val * se)

# (a) Region 1
ci_r1 = confidence_interval(region1)
print("99% CI for Region 1 Flavanoids:", ci_r1)

# (b) Region 3
ci_r3 = confidence_interval(region3)
print("99% CI for Region 3 Flavanoids:", ci_r3)

# (c) Overlap?
overlap = not (ci_r1[1] < ci_r3[0] or ci_r3[1] < ci_r1[0])
print("Do intervals overlap?", overlap)
```
