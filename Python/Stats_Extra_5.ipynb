```python
import numpy as np
import math

# Parameters
simulations = 1_000_000 
n = 10  

# (a) Probability that exactly two people share a birthday
successes = 0
for _ in range(simulations):
    birthdays = np.random.randint(1, 366, n) 
    unique, counts = np.unique(birthdays, return_counts=True)
    # Check if exactly two people share a birthday
    if np.sum(counts == 2) == 1 and np.all(counts <= 2):
        successes += 1

simulated_probability = successes / simulations
print(f"Simulated Probability (n=10): {simulated_probability:.6f}")

# (b) Exact probability calculation for exactly two people sharing a birthday
combinations = math.comb(n, 2)  
shared_birthday = 1 / 365 

# Probability that the remaining people have different birthdays
different_birthdays = 1
for i in range(n - 2):
    different_birthdays *= (364 - i) / 365

exact_prob = combinations * shared_birthday * different_birthdays
print(f"Exact Probability (n=10): {exact_prob:.6f}")
print(f"Difference: {abs(simulated_probability - exact_prob):.6f}")

# (c) Find minimum number of people for probability > 20%
target_prob = 0.20
n = 2
while True:
    combinations = math.comb(n, 2)
    shared_birthday = 1 / 365
    different_birthdays = 1
    for i in range(n - 2):
        different_birthdays *= (364 - i) / 365
    probability = combinations * shared_birthday * different_birthdays

    if probability > target_prob:
        print(f"Minimum people for probability > 20%: {n} (Probability: {probability:.6f})")
        break
    n += 1
```
