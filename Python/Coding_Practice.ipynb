**Task1:**

Write a program that calculates the sum of all odd numbers between two numbers provided by the user.

Instructions:

- Prompt the user for the starting and ending numbers of the range.
- Use a loop to iterate over all numbers in the given range.
- Add only the odd numbers to a cumulative sum.
- Print the resulting sum.



```python
start = int(input("Enter your first number"))
end = int(input("Enter your second number"))

sum = 0

for num in range(start,end+1):
    if num % 2 != 0:
        sum = sum+num

print("The sum of the odd numbers between",start,"and",end,"is",sum)
    
```

**Task:2**

Create a simple number guessing game.

- The program should choose a random number between 1 and 100.
- Allow the user to guess the number until they get it correct.
- Provide hints like "low" or "high" to guide the user.



```python
import random
guesses = 0
secret_number = random.randint(1, 100)
while True:
    
    guess = int(input("Guess a number between 1 and 100: "))
    guesses = guesses + 1

    # Check if the guess is correct
    if guess == secret_number:
        print("Congratulations! You guessed the number in", guesses, "tries.")
        break
    elif guess < secret_number:
        print("Too low! Try again.")
    else:
        print("Too high! Try again.")

                  
```

**Task3:**

Write a program that takes an integer and a digit (0-9) as input from the user and counts how many times that digit appears in the integer.

Instructions:

- Prompt the user to enter an integer.
- Prompt the user to enter the digit to search for.
- Use a loop to extract each digit from the integer.
- Keep a counter to track how many times the digit occurs.
- Print the resulting count.

**Hints:**

Use % and // to extract each digit.

```python
num = int(input("Enter a number to analyze"))
digit = int(input("Which digit would you analyze"))

counter = 0

num_str = str(num)

for n in num_str:
    if int(n) == digit:
        counter = counter + 1

print("The digit",digit,"is in",num,",",counter,"times")

          
```

**Task:4**

Write a program that calculates the sum of a given series: 1 + 1/2 + 1/3 + ... + 1/n, where n is provided by the user.

Instructions:

- Prompt the user to enter a positive integer n.
- Use a loop to calculate the sum of the series up to 1/n.
- Print the resulting sum.

**Hints:**

- Use a loop to iterate from 1 to n.
- Use += to add 1/i to the sum.



```python
n = int(input("Enter a number"))

sum = 0

for i in range(1,n+1):
    sum = sum + (1/i)

print("The sum of the series is: ",sum)
```

**Task:5**

You are provided with a list of numbers representing sales data for different months. Your goal is to determine:

- The maximum sale
- The minimum sale
- The average sale

**Instructions:**

Write a program to determine the above values in two different ways:

- Using Python built-in functions: Use functions like max(), min(), and sum() to calculate the values.

- Using loops: Write loops to calculate each value manually.

```python
sales_data = [1500, 2700, 2200, 3400, 1200, 3100, 2900, 3300, 1800, 4000]

maximum_sale = max(sales_data)
minimum_sale = min(sales_data)
average_sale = sum(sales_data) / len(sales_data)

print("Maximum sale:", maximum_sale)
print("Minimum sale:", minimum_sale)
print("Average sale:", average_sale)

maximum_sale = sales_data[0]  
minimum_sale = sales_data[0] 

total_sales = 0

for sale in sales_data:
    if sale > maximum_sale:
        maximum_sale = sale
    if sale < minimum_sale:
        minimum_sale = sale
    total_sales += sale

average_sale = total_sales / len(sales_data)

print("Maximum sale:", maximum_sale)
print("Minimum sale:", minimum_sale)
print("Average sale:", average_sale)
```
