```python
import numpy as np

# Problem 2b

data = np.array([50.1, 70.1, 137.0,166.9, 170.5, 152.8, 80.5, 123.5, 112.6, 148.5, 160.0, 125.4, 110.6, 99.5]) 

sample_mean = np.mean(data)

lambda_hat = 1/sample_mean

print("The Method of Moments Estimate of Lambda:", lambda_hat)
```

```python
#Problem 3c

data2 = np.array([22.0, 23.9, 20.9, 23.8, 25.0, 24.0, 21.7, 23.8, 22.8, 23.1, 23.1, 23.5, 23.0, 23.0, 21.0])

new_mean = np.mean(data2)
variance = np.var(data2, ddof=0)

alpha_hat = (new_mean**2)/variance
beta_hat = variance/new_mean

print("Method of Moments Estimate of alpha:", alpha_hat)
print("Method of Moments Estimate of beta:", beta_hat)
```
